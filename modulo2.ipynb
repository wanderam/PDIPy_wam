```python
import numpy as np
```

#### while

```python
# Programa para adicionar um número a uma lista

lista = list(np.random.rand(100))

print(lista)
```

```python
n_elementos = len(lista)
print(f'Quantos elementos há nessa lista? {n_elementos}')
```

```python
# Aplicando o while

lista2 = []
contador = 0

while contador < len(lista):
    lista2.append(lista[contador] + 1)
    contador += 1
```

```python
print(lista2)
```

#### for

```python
# Criando uma nova lista
"""
Existe uma convenção em programação que diz para denominarmos o "elemento" por apenas "i".
Isso serve para deixar o código mais limpo.
Ficando:
for i in lista:
    faça isso

Em vez disso:

for elemento in lista:
    lista3.append(elemento + 1)

Faça isso:

for i in lista:
    lista3.append(i + 1)
"""

lista3 = []

for i in lista:
    lista3.append(i + 1)
```

```python
print(lista3)
```

#### Iterando sobre uma lista de listas (Arrays) usando loop for

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])

matriz
```

```python
for i in matriz:
    print(i)
```

```python
# Pense em camadas:
# sendo a primeira camada a lista maior
# e a segunda camada são as duas listas dentro da lista maoir.

for i in matriz:
    for j in i:
        print(j)
```

```python
# Criando uma matriz de zeros no formato da matriz m

matriz2 = np.zeros_like(matriz)
matriz2
```

```python
for i in matriz:
    for j in i:
        matriz2[:j] = matriz[:j] + 1
```

```python
matriz2
```

## Aula 8: Funções

```python
# Exemplo prático e simples de função
# Uma função que calcula a soma de três números

def soma(n1, n2, n3):
    s = n1 + n2 + n3
    return s
```

```python
soma(3.14, 2.75, 8)
```

```python
# Função com parâmetros pré-definidos

def soma2(num1 = 10, num2 = 20, num3 = 30):
    s = num1 + num2 + num3
    return s
```

```python
soma2(15, 20, 45)
```

```python
lista_n = list(np.random.rand(100))
```

```python
# Adicionando um valor ao elemento de uma lista usando for loop
# dentro de uma função


def somma(l, n):
    lista_vazia = []
    for i in l:
        lista_vazia.append(i + n)
    return lista_vazia
```

```python
resultado_soma = somma(lista_n, 3)
print(resultado_soma)
#### Vetorização do Numpy
```

## Aula 9: Programação Orientada a Objetos (POO)

Método é uma ação que é executado em um objeto que pertence a uma classe.

```python
# Criando a classe "Satelite"

class Sat:
    def __init__(self, carga, sensor, velocidade):
        self.carga = carga
        self.sensor = sensor
        self.velocidade = velocidade

    # Criando um método para acelerar e desacelerar o satélite
    def acelera(self, v1):
        self.velocidade += v1

    def desacelera(self, v1):
            self.velocidade -= v1
```

```python
wam_sat = Sat(5000, 'MSS', 2000)
print(wam_sat)
```

```python
wam_sar = Sat(6000, 'SAR-C', 3000)
```

```python
# Printando os objetos para verificar que são diferentes,
# os endereços indicam que estão armazenados em compartimenos diferentes
# logo são objetos diferentes

print(wam_sat, '\n', wam_sar)
```

```python
wam_sat.velocidade
```

```python
wam_sat.acelera(100)
wam_sat.velocidade
```

```python
wam_sat.velocidade
```

```python
wam_sat.desacelera(100)
wam_sat.velocidade
```

#### Entendo melhor os métodos de classes utilizando Numpy

```python
M = np.array([[1, 2, 3], [4, 5, 6]])

M
```

```python
# O Numpy já tem métodos para os respectivos objetos

M.transpose()
```

```python
# Abordagem do método do Numpy
M.sum()
```

```python
# Aplicando a abordagem de função

def sum_elem(l):
    soma = 0
    for i in l:
        for j in i:
            soma += j
    return soma
```

```python
# O resultado é o mesmo
sum_elem(M)
```
